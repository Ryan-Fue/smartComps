# Embedding and Normalization Validation
This notebook tests the `FeatureProcessor` from `src/embedder.py` on the cleaned dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
import sys

# Add src to path
sys.path.append('../')
from src.embedder import FeatureProcessor
```

```python
# Load data
data_path = "../data/processed/company_metrics_clean.parquet"

df = pd.read_parquet(data_path)
print(f"Loaded {len(df)} rows")
display(df.head())
```

## 1. Test Normalization
Using `RobustScaler` to handle outliers in financial data.

```python
processor = FeatureProcessor()
df_norm = processor.normalize_metrics(df.head(100)) # Small sample for quick test

cols_to_check = ["forwardPE", "ev_to_ebitda", "ebitda_margin"]

fig, axes = plt.subplots(1, 2, figsize=(15, 5))
sns.boxplot(data=df[cols_to_check], ax=axes[0]).set_title("Before Scaling")
sns.boxplot(data=df_norm[cols_to_check], ax=axes[1]).set_title("After Scaling (RobustScaler)")
plt.show()
```

## 2. Test Embedding
Using `all-MiniLM-L6-v2` local model.

```python
df_emb = processor.embed_summaries(df.head(10)) # Small sample

display(df_emb.iloc[:,12:396].head())
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# Calculate similarity between first two companies
sim = cosine_similarity([df_emb.loc[0,"nlp_0":"nlp_383"]], [df_emb.loc[1,"nlp_0":"nlp_383"]])
print(f"Similarity between {df_emb.iloc[0]['ticker']} and {df_emb.iloc[1]['ticker']}: {sim[0][0]:.4f}")
```
